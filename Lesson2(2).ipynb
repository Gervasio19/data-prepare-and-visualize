Compute: Rank the US states and territories by their 2010 population density.

Code

```python
import pandas as pd

# 1. Đọc dữ liệu (code của bạn)
url1 = 'https://raw.githubusercontent.com/jakevdp/data-USstates/master/state-population.csv'
url2 = 'https://raw.githubusercontent.com/jakevdp/data-USstates/master/state-areas.csv'
url3 = 'https://raw.githubusercontent.com/jakevdp/data-USstates/master/state-abbrevs.csv'

pop = pd.read_csv(url1)
areas = pd.read_csv(url2)
abbrevs = pd.read_csv(url3)

# 2. Gộp pop và abbrevs dựa trên cột viết tắt
merged = pd.merge(pop, abbrevs, how='outer', left_on='state/region', right_on='abbreviation')
merged = merged.drop('abbreviation', axis=1) # Bỏ cột trùng lặp cho gọn

# 3. Xử lý các tên bang bị thiếu (NaN) do không có trong bảng abbrevs
merged.loc[merged['state/region'] == 'PR', 'state'] = 'Puerto Rico'
merged.loc[merged['state/region'] == 'USA', 'state'] = 'United States'

# 4. Lọc dữ liệu năm 2010 và tổng dân số (ages == 'total')
data_2010 = merged[(merged['year'] == 2010) & (merged['ages'] == 'total')]

# 5. Gộp với bảng diện tích (areas)
final_data = pd.merge(data_2010, areas, on='state', how='left')

# Loại bỏ dòng của toàn bộ nước Mỹ (United States) vì ta chỉ muốn xếp hạng các bang/vùng lãnh thổ
final_data = final_data[final_data['state'] != 'United States']

# 6. Tính mật độ dân số
final_data['density'] = final_data['population'] / final_data['area (sq. mi)']

# 7. Sắp xếp để tạo bảng xếp hạng (từ cao xuống thấp)
ranked_states = final_data.sort_values(by='density', ascending=False)

# Xem kết quả 5 bang có mật độ cao nhất
print(ranked_states[['state', 'density', 'population', 'area (sq. mi)']].head())
```
