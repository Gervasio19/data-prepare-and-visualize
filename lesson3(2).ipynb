Birthrate data

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load the data
url = 'https://raw.githubusercontent.com/jakevdp/data-CDCbirths/master/births.csv'
birth = pd.read_csv(url)

# 2. Create the pivot table (Total births by year and gender)
df = pd.pivot_table(data=birth,
                    index='year',
                    columns='gender',
                    values='births',
                    aggfunc='sum')

# 3. Generate the plots

# --- Plot 1: Line (Default) ---
# Best for viewing continuous trends over time.
df.plot(kind='line', title='Yearly Births by Gender (Line)', 
        xlabel='Year', ylabel='Total Births')
plt.show()

# --- Plot 2: Bar ---
# Best for comparing discrete categories. (Note: With 20+ years, 
# a bar chart can look quite dense compared to a line chart!)
df.plot(kind='bar', title='Yearly Births by Gender (Bar)', 
        xlabel='Year', ylabel='Total Births')
plt.show()

# --- Plot 3: Histogram ---
# Best for seeing the distribution of birth counts.
# (e.g., "How many years had roughly 1.9 million male births?")
df.plot(kind='hist', title='Distribution of Birth Counts (Histogram)', 
        xlabel='Number of Births', ylabel='Frequency (Years)', alpha=0.7)
plt.show()

# --- Plot 4: Boxplot ---
# Best for statistical spread, showing the median, quartiles, and any outliers.
df.plot(kind='box', title='Statistical Spread of Births (Box)', 
        xlabel='Gender', ylabel='Total Births')
plt.show()

# --- Plot 5: Scatter ---
# Best for showing correlation. Here, it shows that as female 
# births increase, male births increase proportionally.
df.plot(kind='scatter', x='F', y='M', 
        title='Female vs Male Births Correlation (Scatter)', 
        xlabel='Female Births', ylabel='Male Births')
plt.show()
```
