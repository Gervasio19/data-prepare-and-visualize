```python
planets[['method','number']].groupby('method').sum().sort_values(by='number',ascending=False)
```

```python
#sum of number, means of mass, min and max of orbital_period
planets.groupby('year').agg(
    sum=('number','sum'),
    avg_mass=('mass','mean'),
    min_orbit=('orbital_period', 'min')
```

```python
planets.groupby('year').agg(
    tong_so_hanh_tinh=('number', 'sum'),
    khoi_luong_tb=('mass', 'mean'),
    quy_dao_ngan_nhat=('orbital_period', 'min'),
    cac_phuong_phap=('method', lambda x: list(x.unique())) # Danh sách các method dùng trong năm
)
```
