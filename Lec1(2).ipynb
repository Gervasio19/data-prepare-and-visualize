```python
import pandas as pd

# Dữ liệu ban đầu
data = {
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Score': [88, 76, 95, 65, 70],
    'Days_Off': [2, 5, 1, 8, 6]
}
df_lab1 = pd.DataFrame(data)

# Yêu cầu 1: Định nghĩa hàm đánh giá hiệu suất dựa trên Score và Days_Off
def evaluate_performance(row):
    score = row['Score']
    days_off = row['Days_Off']

    if score >= 85 and days_off <= 3:
        return 'Excellent'
    elif score >= 70 and days_off <= 5:
        return 'Good'
    elif score >= 50:
        return 'Average'
    else:
        return 'Poor'

# Yêu cầu 2: Áp dụng hàm cho từng hàng (row) sử dụng .apply()
# Lưu ý: axis=1 để áp dụng hàm trên từng hàng
df_lab1['Performance'] = df_lab1.apply(evaluate_performance, axis=1)


# Yêu cầu 3: Định nghĩa hàm phân loại Score thành "High" hoặc "Low"
# (Giả sử mốc là 80, bạn có thể tự chỉnh sửa)
def categorize_score(score):
    if score >= 80:
        return 'High'
    else:
        return 'Low'

# Yêu cầu 4: Áp dụng hàm thứ hai sử dụng .map() cho cột Score
df_lab1['Score_Category'] = df_lab1['Score'].map(categorize_score)

# In kết quả để kiểm tra
print(df_lab1)
```
