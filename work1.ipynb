How do revenue, profit, employment, and labor productivity differ among state-owned, domestic non-state, and foreign-invested enterprises?

```python
df_2019_core['LoaihinhKT'] = pd.to_numeric(
    df_2019_core['LoaihinhKT'],
    errors='coerce'
)
```

```python
def classify_ownership(x):
    if 1 <= x <= 4:
        return 'State-owned'
    elif 5 <= x <= 10:
        return 'Domestic non-state'
    elif 11 <= x <= 13:
        return 'Foreign-invested'
    else:
        return 'Unknown'
```

```python
# Create a new variable
df_2019_core['Ownership_group'] = (
    df_2019_core['LoaihinhKT']
    .apply(classify_ownership)
)
```

```python
df_2019_core['Ownership_group'].value_counts()
```

```python
df_2019_core.groupby('Ownership_group').agg(
    Avg_revenue=('Doanhthu_Nganh', 'mean'),
    Avg_profit=('Loinhuan_SauThue', 'mean'),
    Avg_employment=('SoLaodong_CuoiNam', 'mean')
)
```

```python
Use groupby().transform() to calculate the average revenue within each ownership group.
Create a new variable indicating whether each enterprise has revenue above or below its group average.
```

```python
# Calculate the average revenue for each ownership group
df_2019_core['Group_avg_revenue'] = (
    df_2019_core
    .groupby('Ownership_group')['Doanhthu_Nganh']
    .transform('mean')
)

df_2019_core
```

```python
# Compare each firm's revenue with the average revenue of its ownership group
df_2019_core['Above_group_avg_revenue'] = (
    df_2019_core['Doanhthu_Nganh']
    > df_2019_core['Group_avg_revenue']
)
```

```python
df_2019_core['Above_group_avg_revenue'].astype(int)
```

```python
# Calculate the difference between firm revenue
# and the average revenue of its ownership group
df_2019_core['Revenue_gap'] = (
    df_2019_core['Doanhthu_Nganh']
    - df_2019_core['Group_avg_revenue']
)
```
