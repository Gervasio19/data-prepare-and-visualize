# Exercise 2. LOF – Reachability, LRD and LOF

Use:

- $k\text{-dist}(A)=3.00$
- $k\text{-dist}(B)=4.24$
- $k\text{-dist}(C)=4.24$
- $k\text{-dist}(P)=15.00$

$$
rd_k(p,o)=\max\{k\text{-dist}(o),d(p,o)\}
$$

$$
LRD_k(p)=\frac{|N_k(p)|}{\sum_{o\in N_k(p)}rd_k(p,o)}
$$

$$
LOF_k(p)=\frac{1}{|N_k(p)|}\sum_{o\in N_k(p)}\frac{LRD_k(o)}{LRD_k(p)}
$$

```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer

# Tạo dữ liệu mẫu chứa các giá trị quan sát và một giá trị khuyết (np.nan)
data = np.array([[700], [680], [640], [710], [np.nan]])

# Khởi tạo SimpleImputer với chiến lược tính 'mean'
imputer = SimpleImputer(strategy='mean')

# Fit mô hình và biến đổi dữ liệu
imputed_data = imputer.fit_transform(data)

# In giá trị trung bình mà SimpleImputer đã học được
print(f"Giá trị trung bình được tính bởi SimpleImputer: {imputer.statistics_[0]}")

# Hiển thị mảng sau khi đã điền giá trị khuyết
print("\nDữ liệu sau khi điền khuyết thiếu:")
print(imputed_data)
```
