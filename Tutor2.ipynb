A2. Harmonize variables across years

```python
df_2019 = df_2019_raw.copy()

core_cols = [
    'Masothue', 'TenDN', 'LoaihinhKT', 'MaNganhSXKD_Chinh', 'MaTinh_Dieutra', 
    'TongDoanhthu_DTDN', 'Loinhuan_SauThue', 'SoLaodong_CuoiNam', 'SoLaodong_BQ'
]

df_2019 = df_2019[core_cols]

df_2019['Year'] = 2019

```

PART B

B1

```python
missing_report = pd.DataFrame({
    'Missing_Count': df_2019.isnull().sum(),
    'Missing_Percentage': (df_2019.isnull().sum() / len(df_2019)) * 100,
    'Data_Type': df_2019.dtypes
})

missing_report = missing_report.sort_values(by='Missing_Percentage', ascending=False)
display(missing_report)

```

B2

```python
exact_duplicates = df_2019.duplicated().sum()
print(f"Exact duplicates count: {exact_duplicates}")

mst_duplicates = df_2019.duplicated(subset=['Masothue']).sum()
print(f"Duplicated Tax IDs count: {mst_duplicates}")

display(df_2019[df_2019.duplicated(subset=['Masothue'], keep=False)].sort_values('Masothue').head(10))

df_2019 = df_2019.drop_duplicates()

df_2019['Masothue'] = df_2019['Masothue'].astype(str).str.strip()
df_2019['Masothue'] = df_2019['Masothue'].str.replace(r'\.0$', '', regex=True)
```

B3

```python
invalid_labor = df_2019[df_2019['SoLaodong_CuoiNam'] < 0]
print(f"Lao động < 0: {len(invalid_labor)} dòng")

revenue_q1 = df_2019['TongDoanhthu_DTDN'].quantile(0.01)
revenue_q99 = df_2019['TongDoanhthu_DTDN'].quantile(0.99)
print(f"Doanh thu 1% thấp nhất: {revenue_q1}")
print(f"Doanh thu 99% cao nhất: {revenue_q99}")

top_10_revenue = df_2019.nlargest(10, 'TongDoanhthu_DTDN')
print("\n--- Top 10 Doanh thu ---")
display(top_10_revenue[['Masothue', 'TenDN', 'TongDoanhthu_DTDN']])

df_2019['Abs_Loinhuan'] = df_2019['Loinhuan_SauThue'].abs()
top_10_abs_profit = df_2019.nlargest(10, 'Abs_Loinhuan')
print("\n--- Top 10 Trị tuyệt đối Lợi nhuận ---")
display(top_10_abs_profit[['Masothue', 'TenDN', 'Loinhuan_SauThue', 'Abs_Loinhuan']])

```
