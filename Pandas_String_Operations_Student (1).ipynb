## 5. Regular Expressions (Regex)

A regular expression is a **pattern that describes the text we want to find**. Common symbols include:

| Symbol | Meaning |
|---|---|
| `^` | start of a string |
| `$` | end of a string |
| `.` | any single character |
| `*` | zero or more repetitions |
| `+` | one or more repetitions |
| `\d` | one digit |
| `\s` | one whitespace character |
| `[A-Za-z]` | one English letter |
| `[^AEIOU]` | one character not included in the specified set |
| `|` | OR |
| `()` | a capturing group |

## 6. Combining Multiple Search Conditions with `|`

In regex, `|` means **OR**. The expression `'|'.join(keyword_list)` converts a list of keywords into one regex pattern.

```python
df_2019_core = pd.DataFrame({
    'Masothue': ['0101', '0102', '0103', '0104', '0105', '0106', '0107'],
    'TenDN': [
        '  Công ty cổ phần Vingroup ',
        'CÔNG TY CỔ PHẦN VINHOMES',
        'Công ty TNHH Sản xuất VinFast',
        'Công ty Vinpearl Nha Trang',
        'Công ty May Việt Tiến',
        'Công ty Vinamilk',
        None
    ]
})

df_2019_core['TenDN_clean'] = (
    df_2019_core['TenDN']
    .str.strip()
    .str.upper()
    .str.replace(r'\s+', ' ', regex=True)
)

df_2019_core
```
